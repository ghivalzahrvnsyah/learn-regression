```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
df_train = pd.read_csv("/kaggle/input/competitions/playground-series-s4e5/train.csv")
df_test = pd.read_csv("/kaggle/input/competitions/playground-series-s4e5/test.csv")
df_test
```

```python
df_train.info()
```

```python
## Menampilkan Statistik deskriptif dari dataset

df_train.describe(include="all")
```

```python
missing_values = df_train.isnull().sum()
missing_values[missing_values > 0]
```

```python
for feature in df_train.columns:
    plt.figure(figsize=(10,6))
    sns.boxplot(x=df_train[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
```

```python
## Mengindentifikasi outliers menggunakan IQR

Q1 = df_train.quantile(0.25)
Q3 = df_train.quantile(0.75)
IQR = Q3 - Q1

# Filter dataframe hanya menyimpan baris yang tidak mengandung outliers pada kolom numerik
condition = ~((df_train < (Q1 - 1.5 * IQR)) | (df_train > (Q3 + 1.5 * IQR))).any(axis=1)
df = df_train.loc[condition, df_train.columns]
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x=df['WetlandLoss'])
plt.title(f'Box Plot of WetlandLoss')
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

# Memastikan hanya data numerical yang diproses
numeric_features = df.select_dtypes(include=['number']).columns
numeric_features

# Standarisasi fitur numeric
scaler = StandardScaler()
df[numeric_features] = scaler.fit_transform(df[numeric_features])
```

```python
# Mengidentifikasi duplikat 
duplicates = df.duplicated()

print("Baris duplikat:")
print(df[duplicates])

df.describe(include='all')
```

```python
## Visualisasi standar deviasi

# Menghitung jumlah variable
num_vars = df.shape[1]

# menentukan jumlah baris dan kolom untuk grid subplot
n_cols = 4
n_rows = -(-num_vars)

# membuat plot
fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, n_rows * 4))

# Flatten axes array untuk memudahkan iterasi jika diperlukan
axes = axes.flatten()

# plot setiap variabel
for i, column in enumerate(df.drop(columns=["id"]).columns):
    df[column].hist(ax=axes[i], bins=20, edgecolor='black')
    axes[i].set_title(column)
    axes[i].set_xlabel('Value')
    axes[i].set_ylabel('Frequency')

# Menghapus subplot yang tidak terpakai
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Sesuaikan layout agar rapi 
plt.tight_layout()
plt.show()
```

```python
## Memilih Fitur yang Memiliki Hubungan dengan Fitur Target

# Menghitung korelasi variabel target dengan variabel lainnya
target_corr = df.corr()['FloodProbability']

# Mengurutkan hasil korelasi
target_corr_sorted = target_corr.abs().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
target_corr_sorted.plot(kind='bar')
plt.title(f'Correlation with Flood Probability')
plt.xlabel(f'Variables')
plt.ylabel(f'Correlation Coefficient')
plt.show()
```

```python
## Data Splitting

import sklearn
from sklearn import datasets

# memisahkan fitur (x) dan target (y)
X = df.drop(columns=['FloodProbability'])
y = df['FloodProbability']

from sklearn.model_selection import train_test_split

# membagi dataset ke dalam data training dan data testing
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# menghitung jumlah data
print("jumlah data: ", len(X))

# panjang data x_train
print("Jumlah data: ", len(x_train))

# panjang data x_test
print("Jumlah data: ", len(x_test))
```

```python
## Melatih Model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Menggunakan LARS
from sklearn import linear_model
lars = linear_model.Lars(n_nonzero_coefs=1).fit(x_train, y_train)
pred_lars = lars.predict(x_test)

mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)
```

```python
 # Mengunakan Linear Regression
from sklearn.linear_model import LinearRegression
LR = LinearRegression().fit(x_train, y_train)
 
pred_LR = LR.predict(x_test)
 
mae_LR = mean_absolute_error(y_test, pred_LR)
mse_LR = mean_squared_error(y_test, pred_LR)
r2_LR = r2_score(y_test, pred_LR)
```

```python
# menggunakan gradientBoostingRegressor
from sklearn.ensemble import GradientBoostingRegressor

GBR = GradientBoostingRegressor(random_state=184)
GBR.fit(x_train, y_train)

pred_GBR = GBR.predict(x_test)
mae_GBR = mean_absolute_error(y_test, pred_GBR)
mse_GBR = mean_squared_error(y_test, pred_GBR)
r2_GBR = r2_score(y_test, pred_GBR)
```

```python
# Membuat dictionary untuk menyimpan hasil evaluasi
data = {
    'MAE': [mae_lars, mae_LR, mae_GBR],
    'MSE': [mse_lars, mse_LR, mse_GBR],
    'R2': [r2_lars, r2_LR, r2_GBR]
}
 
# Konversi dictionary menjadi DataFrame
df_results = pd.DataFrame(data, index=['Lars', 'LR', 'GBR'])
df_results
```
